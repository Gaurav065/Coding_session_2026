# Kaggriculture - LSTM Behavioral Cloning

This notebook trains an LSTM network to predict the 30-day macro-strategy (Target Portfolio) of Grandmaster players.
The LSTM structure solves the 'jitter' problem found in standard MLPs by naturally maintaining a memory of its own long-term strategy across the 30-day season.

```python
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
import matplotlib.pyplot as plt

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")
```

```python
# Load Sequence Data
X = np.load('/kaggle/input/kaggriculture-lstm-dataset/X_seq.npy')
Y = np.load('/kaggle/input/kaggriculture-lstm-dataset/Y_seq.npy')

print(f"X Shape: {X.shape} | Y Shape: {Y.shape}")

dataset = TensorDataset(torch.tensor(X, dtype=torch.float32), torch.tensor(Y, dtype=torch.float32))
loader = DataLoader(dataset, batch_size=32, shuffle=True)
```

```python
class LSTMMacroAgentNet(nn.Module):
    def __init__(self, obs_dim=50, hidden_dim=128, act_dim=9, num_layers=2):
        super(LSTMMacroAgentNet, self).__init__()
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers
        
        # The LSTM takes a sequence of observations (Batch, Seq, Obs)
        self.lstm = nn.LSTM(obs_dim, hidden_dim, num_layers, batch_first=True)
        
        # Predict targets for each step in the sequence
        self.fc = nn.Sequential(
            nn.Linear(hidden_dim, 64),
            nn.ReLU(),
            nn.Linear(64, act_dim),
            nn.Sigmoid()
        )
        
    def forward(self, x, hx=None):
        # x shape: (batch_size, seq_length, obs_dim)
        out, hidden = self.lstm(x, hx)
        # out shape: (batch_size, seq_length, hidden_dim)
        pred = self.fc(out)
        return pred, hidden

model = LSTMMacroAgentNet().to(device)
optimizer = optim.Adam(model.parameters(), lr=0.001)
criterion = nn.MSELoss()
```

```python
epochs = 200
losses = []

for epoch in range(epochs):
    epoch_loss = 0
    for batch_x, batch_y in loader:
        batch_x, batch_y = batch_x.to(device), batch_y.to(device)
        
        optimizer.zero_grad()
        preds, _ = model(batch_x) # We pass the whole 30-day sequence at once
        
        loss = criterion(preds, batch_y)
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item()
        
    avg_loss = epoch_loss / len(loader)
    losses.append(avg_loss)
    if (epoch+1) % 10 == 0:
        print(f"Epoch {epoch+1}/{epochs} | Loss: {avg_loss:.6f}")
```

```python
plt.plot(losses)
plt.title('LSTM Behavioral Cloning Loss')
plt.xlabel('Epoch')
plt.ylabel('MSE Loss')
plt.show()
```
